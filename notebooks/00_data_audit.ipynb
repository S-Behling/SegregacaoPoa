```python
from pathlib import Path

import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
PROJECT_ROOT = Path.cwd().parent

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

print(PROCESSED_DIR)
```

```python
for arquivo in PROCESSED_DIR.iterdir():
    print(arquivo.name)
```

```python
renda_setores = pd.read_csv(
    PROCESSED_DIR / "renda_setores_poa.csv",
    dtype={"CD_SETOR": str}
)

setores = gpd.read_file(
    PROCESSED_DIR / "setores_poa.gpkg",
    layer="setores_poa"
)

bairros = gpd.read_file(
    PROCESSED_DIR / "bairros_poa.gpkg",
    layer="bairros_poa"
)

limite = gpd.read_file(
    PROCESSED_DIR / "limite_poa.gpkg",
    layer="limite_poa"
)
```

```python
display(renda_setores.head())
display(setores.head())
display(bairros.head())
display(limite.head())
```
